*  The Easy Set

* The Hard Set
Multiple fonts, fluctuating capitalization across individual letters, and noisy or textured
backgrounds. Ensure diversity in the dataset to test your model’s ability to generalize.


* The Bonus Set
This set is used for the bonus Generation task and can be ignored if you intend to skip that task.
It borrows all conditions from the hard set with the added condition that if the background is
green, the word is rendered normally, but if the background is red, the word is rendered in
reverse. Note: The output does not change. For example, if “hello” is rendered on a red image
as “olleh,” your model should still produce “hello.”

```python
import random
from PIL import Image, ImageDraw, ImageFont
import numpy as np
import os
```

```python
os.makedirs("dataset/bonus", exist_ok=True)

words = ["hello", "world", "india", "captcha", "vision", "text", "image","jhansi","patel","kashmir","pen","book","taj","tea","sugar"]
fonts = ["/content/Arial.ttf", "/content/TimesNewRoman.ttf", "/content/CourierNew.ttf"]
```

```python

def random_case(word):
    return "".join([c.upper() if random.random()>0.5 else c.lower() for c in word])
```

```python
def add_noise(img):
    arr = np.array(img)
    noise = np.random.randint(0,30,(arr.shape[0], arr.shape[1], arr.shape[2]))
    arr = arr + noise
    arr = np.clip(arr,0,255)
    return Image.fromarray(arr.astype('uint8'))
```

```python

for w in words:
    for i in range(100):
        font = ImageFont.load_default()
        word_mod = random_case(w)

        # Random background
        bg_color = random.choice(["green","red"])
        img = Image.new("RGB", (200,60), color=bg_color)
        draw = ImageDraw.Draw(img)

        # Red background → reverse word for rendering
        display_word = word_mod[::-1] if bg_color=="red" else word_mod
        draw.text((10,10), display_word, font=font, fill=(0,0,0))

        img = add_noise(img)
        img.save(f"dataset/bonus/{w}_{i}.png")
```
